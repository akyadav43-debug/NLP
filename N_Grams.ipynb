***N-grams***

*  N=grams preserve the seqential structure of language, aiding in context understanding and feature extraction.



# Types

*  Unigram- single words
*  Biagram-pairs of consecutive words.
*  Trigram=triplet of consecutive words.







```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.util import ngrams
nltk.download('punkt_tab')
```

```python
def generate_ngrams(text, n):
    token = word_tokenize(text)
    n_grams = list(ngrams(token, n))
    return n_grams
```

```python
##Example text
txt ='N-Grams are a sequence of n items from a given sample of text or speech'
unigram= generate_ngrams(txt,1)
bigram= generate_ngrams(txt,2)
trigram= generate_ngrams(txt,3)
print(unigram)
print(bigram)
print(trigram)
```
