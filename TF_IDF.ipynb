TF-IDF (Term frequency- inverse Document Frequency)--
A numerical statistics that reflects the importance of a tern=m on a document relative to its occurence across multiple documents. If the value of varience is closer to mean then it's a low varience.

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus =[
    "Thor is eating pizza, Loki is eating Pizza, Ironman ate pizza already",
    "Apple is announcing new Iphone tomorrow",
    "Tesla is announcing new model-3 tomorrow",
    "Google is announcing new pixel-6 tomorrow",
    "Amazon is announcing new eco-dot tomorrow",
    "I am eating biryani and you are eating grapes"
]
```

```python
v= TfidfVectorizer()
v.fit(corpus)
transform_output=v.transform(corpus)
```

```python
print(v.vocabulary_)
```

```python
##print IDF of each word
all_feature_names= v.get_feature_names_out()
for word in all_feature_names:
    indx=v.vocabulary_.get(word)
```

```python
###get the score
idf_score=v.idf_[indx]
print(f"{word}: {idf_score}")
```

```python
###Print transformed output from tf-idf
print(transform_output.toarray())
```
