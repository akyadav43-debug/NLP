##Libraries in NLP


-spacy - Natural language processing library in python that can be tokennized and process textual data

NLTK

```python
## import spacy
import spacy
#Load spacy Model
nlp= spacy.load("en_core_web_sm")

##input string
input_str= "Hello How@ are you!"

##Function to clean the string
def clean_string(text):
  cleaned_text=''.join(char for char in text if char.isalpha() or char.isspace())
  doc=nlp(cleaned_text)
  return ' '.join(token.text for token in doc)
##get final output
clean_str= clean_string(input_str)
print(clean_str)
```

```python
import nltk
nltk.download("punkt_tab")
```

```python
input_str= "Hello How@ are you!"

##Tokenize
tokens = nltk.word_tokenize(input_str)
print(tokens)

##REmove the special character

clean_tokens =[token for token in tokens if token.isalnum()]
clean_str= ' '.join(clean_tokens)
print(clean_str)
```
