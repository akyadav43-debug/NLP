# Word Embedding
---
*   What are word embeddigs?
*   Word Embeddings are numerical representation of words that capture semantic relationship and contextual similarities.


---


**Key concepts**
*   dense Vector represantation: Words are represented as high-dimensional vectors with continuous values.
*   Semantic Similarities: Words with similar meanings have similar vector representations.
*   Context Understandings: Captures relationships and context between words.








# Bag-of-words(BoW)
*   A representation of text as an unordered set of words, simplifying language for analysis.

---
**Importance in NLP**


*  Simplified Representation:BoW transforms text into a numerical format, enabling easy application of machine learning algorithms.
