```python
import re
def expand_contractions(text):
  contractions_pattern = {
      r"ain't": "is not",
      r"aren't": "are not",
      r"can't": "cannot",
      r"can't've": "cannot have",
      r"'cause": "because",
      r"could've": "could have",
      r"couldn't": "could not",
      r"couldn't've": "could not have",
      r"didn't": "did not",
      r"doesn't": "does not",
      r"don't": "do not",
      r"hadn't": "had not",
      r"hadn't've": "had not have",
  }
  for contraction, expansion in contractions_pattern.items():
    text = re.sub(contraction, expansion, text)
  return text
```

```python
txt= "I couldn't believe it's raining"
expanded_txt=expand_contractions(txt)
print(expanded_txt)
```

NLP-TOkenization

*   sentence Tokenization
*   Word Tokenization



```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize, word_tokenize
##Sample text for tokenization
txt= "NLTK provides powerful tools for tokenization. It includes word and sentence tokenization."
```

```python
###word tokenize
word=word_tokenize(txt)
print(word)
```

```python
###sentence  tokenize
word=sent_tokenize(txt)
print(word)
```
