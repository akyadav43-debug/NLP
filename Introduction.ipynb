*NLP-Natural Language Processing*

Natural language processing (NLP) is a field of study that focuses on the interaction between the computers and human being(Language).
It involves the development of algorithms and models to enable computers to understand, interpret, and generate human language.

Its applications
Language translation
Sentiment analysis
Chatbots

Key challenges in NLP

1.   Ambiguity-multiple meanings of words
2.   Lack of Standarization: Variation in Languages Across Regions and Dialects
3.   Context Understanding: Grasping Contextual Naunces in Languages
4.   Data starsity: Limited Data for Training Models, Especially in Specified Domains
